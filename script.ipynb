```python
import requests
import time

URL = "https://proof.ovh.net/files/1Gb.dat"

# Limit to approximately 4 MB/s
LIMIT_MBPS = 2
CHUNK_SIZE = 64 * 1024  # 64 KB

delay = CHUNK_SIZE / (LIMIT_MBPS * 1024 * 1024)

while True:
    try:
        print("Downloading...")
        with requests.get(URL, stream=True, timeout=10) as r:
            r.raise_for_status()

            for chunk in r.iter_content(chunk_size=CHUNK_SIZE):
                if not chunk:
                    break
                time.sleep(delay)

    except KeyboardInterrupt:
        print("Stopped.")
        break

    except Exception as e:
        print(f"Error: {e}")
        time.sleep(5)
```
